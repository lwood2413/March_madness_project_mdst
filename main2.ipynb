Adding Data from last ten years, excluding 2019-2020(COVID), Also adding team ratings data. Then, Merge datasets into master_team_stats

```python
import pandas as pd 
import csv

ratings_data = pd.read_csv('538 Ratings.csv')
team_data_25_26 = pd.read_csv('25-26_data.csv')
team_data_24_25 = pd.read_csv('24-25.csv')
team_data_23_24 = pd.read_csv(' 23-24.csv')
team_data_22_23 = pd.read_csv(' 22-23.csv')
team_data_21_22 = pd.read_csv('21-22.csv')
team_data_20_21 = pd.read_csv(' 20-21.csv')
team_data_18_19 = pd.read_csv(' 18-19.csv')
team_data_17_18 = pd.read_csv('17-18.csv')
team_data_16_17 = pd.read_csv(' 16-17.csv')
team_data_15_16 = pd.read_csv(' 15-16.csv')

team_data_25_26['Season'] = '2025-26'
team_data_24_25['Season'] = '2024-25'
team_data_23_24['Season'] = '2023-24'
team_data_22_23['Season'] = '2022-23'
team_data_21_22['Season'] = '2021-22'
team_data_20_21['Season'] = '2020-21'
team_data_18_19['Season'] = '2018-19'
team_data_17_18['Season'] = '2017-18'
team_data_16_17['Season'] = '2016-17'
team_data_15_16['Season'] = '2015-16'

all_seasons = [
    team_data_15_16, team_data_16_17, team_data_17_18, team_data_18_19,
    team_data_20_21, team_data_21_22, team_data_22_23, team_data_23_24,
    team_data_24_25, team_data_25_26
]

master_team_stats = pd.concat(all_seasons, axis=0, ignore_index=True)

master_team_stats.head(10)
```

Clean the data

```python
df = master_team_stats
df = df.rename(columns={'Unnamed: 1': 'School' })
df = df[df["School"].str.contains("NCAA", case=False, na=False)]

df = df.rename(columns={'Overall': 'Games', 'Overall.1': 'Wins', 'Overall.2': 'Losses', 'Overall.3': 'W-L %', 'Overall.3': 'W-L %',
'Overall.4': 'SRS', 'Overall.5': 'Strength of Schedule', 'Conf.': 'Conf. Wins', 'Conf..1': 'Conf. Losses', 'Home': 'Home Wins', 'Home.1':'Home Losses',
'Away': 'Away Wins', 'Away.1': 'Away Losses', 'Points.1': 'Opp. Points', 'Totals': 'Minutes Played', 'Totals.1': 'FG', 'Totals.2': 'FG Att.', 'Totals.3': 'FG %',
 'Totals.4': '3 Pts', 'Totals.5': '3pt att.', 'Totals.6': '3pt %', 'Totals.7': 'FT', 'Totals.8': 'FT att.', 'Totals.9': 'FT %', 'Totals.10': 'Offensive RB', 
 'Totals.11': 'Total RB', 'Totals.12': 'Assists', 'Totals.13': 'Steals', 'Totals.14': 'Blocks', 'Totals.15': 'Turnovers', 'Totals.16': 'PF'})

df = df.drop(columns=[ 'Unnamed: 0', 'Unnamed: 8', 'Unnamed: 11', 'Unnamed: 14', 'Unnamed: 17', 'Unnamed: 20'])


pd.set_option('display.max_columns', None)
df.head(10)

```
